```python
def create_dic(student,score):
  info_dic={
      name : mark for name,mark in zip(student,score)
  }
  return info_dic




students = ["Ali", "Sara", "Ahmed", "Usman", "Hina"]

scores = [78, 92, 65, 88, 95]

attendance = [80, 95, 70, 90, 98]
normal_dic = create_dic(students, scores)
print(normal_dic)

passing_stu=[name for name, marks in normal_dic.items() if marks>=70]
print(passing_stu)
high_performance_stu=[name for name, marks in normal_dic.items()if marks>=70 and attendance[students.index(name)]>=80]
print(high_performance_stu)

```
