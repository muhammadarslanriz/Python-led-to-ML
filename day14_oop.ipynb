```python
models_data = [
    {"name": "ImageClassifier", "framework": "PyTorch", "batch_size": 32,
     "dataset": "Images", "samples": 12000},
    {"name": "TextClassifier", "framework": "TensorFlow", "batch_size": 64,
     "dataset": "Text", "samples": 8000},
    {"name": "FraudDetector", "framework": "PyTorch", "batch_size": -8,
     "dataset": "Transactions", "samples": 50000},
]


class AIModel:
    default_framework = "PyTorch"
    default_batch_size = 32

    def __init__(self, model_name, dataset, samples,
                 framework=None, batch_size=None):
        if framework is None:
            framework = self.default_framework
        if batch_size is None:
            batch_size = self.default_batch_size

        if not isinstance(model_name, str):
            raise ValueError("model name must be a string")
        if not isinstance(framework, str):
            raise ValueError("framework must be a string")
        if not isinstance(batch_size, int) or batch_size <= 0:
            raise ValueError("batch size must be a positive integer")
        if not isinstance(dataset, str):
            raise ValueError("dataset must be a string")
        if not isinstance(samples, int) or samples <= 0:
            raise ValueError("samples must be a positive integer")

        self.model_name = model_name
        self.framework = framework
        self.batch_size = batch_size
        self.dataset = dataset
        self.samples = samples
        self.config = {}
        self.tags = []

    def update_config(self, key, value):
        self.config[key] = value

    def add_tag(self, tag):
        self.tags.append(tag)

    def update_batch_size(self, batch_size):
        if not isinstance(batch_size, int) or batch_size <= 0:
            raise ValueError("batch size must be a positive integer")
        self.batch_size = batch_size

    def describe(self):
        return {
            "model_name": self.model_name,
            "framework": self.framework,
            "batch_size": self.batch_size,
            "dataset": self.dataset,
            "samples": self.samples,
            "config": self.config,
            "tags": self.tags,
        }


models = []
for data in models_data:
    try:
        model = AIModel(
            model_name=data["name"],
            framework=data["framework"],
            batch_size=data["batch_size"],
            dataset=data["dataset"],
            samples=data["samples"],
        )
        models.append(model)
    except ValueError as e:
        print(f"Skipping {data.get('name')}: {e}")

for m in models:
    print(m.describe())
```
