```python
def cal_purshase_stats(customer):
    if not customer:
        return 0, 0, 0, 0

    total = 0
    for pur in customer:
        total += pur

    avg = total / len(customer)
    highest_purchase = max(customer)
    lowest_purchase = min(customer)

    return total, avg, highest_purchase, lowest_purchase


def get_customer_segment(avg_purchase):
    if avg_purchase >= 2000:
        return "premium"
    elif avg_purchase >= 1000:
        return "regular"
    elif avg_purchase >= 500:
        return "low value"
    else:
        return "v low"


def calculate_skill_core(skill):
    score = 0
    required_skills = ["Python", "SQL", "Docker", "AWS"]
    for ski in skill:
        if ski in required_skills:
            score += 1
    persentage = (score / len(required_skills)) * 100
    return persentage, score


def process_customer(customer):
    total, avg, high, low = cal_purshase_stats(customer["purchases"])
    segment = get_customer_segment(avg)
    skill_pct, skill_score = calculate_skill_core(customer["skills"])
    return {
        "name": customer["name"],
        "total_purchase": total,
        "average_purchase": avg,
        "segment": segment,
        "skill_score": skill_score,
        "skill_percentage": skill_pct
    }

def process_all_customers(customers):
    processed_customers = []
    for customer in customers:
        processed_customer = process_customer(customer)
        processed_customers.append(processed_customer)
    return processed_customers


customers = [
    {"name": "Ali",   "age": 25, "purchases": [1200, 800, 450],       "country": "Pakistan", "skills": ["Python", "SQL"]},
    {"name": "Sara",  "age": 31, "purchases": [2500, 1800, 3200, 900], "country": "UAE",     "skills": ["Python", "SQL", "Docker", "AWS"]},
    {"name": "Ahmed", "age": 19, "purchases": [],                     "country": "Pakistan", "skills": ["Java"]},
    {"name": "Hina",  "age": 28, "purchases": [700, 950, 1100],       "country": "Pakistan", "skills": ["Python", "Git"]},
]

processed_customers = [process_customer(customer) for customer in customers]

for customer in processed_customers:
    print(customer)
```
