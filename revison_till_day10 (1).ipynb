```python
records = [
    {"id": 101, "name": "Ali",
     "age": 24, "scores": [85, 90, 78],
     "skills": ["Python", "SQL", "Pandas"]},
    {"id": 102, "name": "Sara",   "age": 22,
     "scores": [92, 88, 95],
     "skills": ["Python", "PyTorch", "SQL"]},
    {"id": 103, "name": "Hamza",
     "age": 31, "scores": [],
     "skills": ["Java"]},
    {"id": 104, "name": "Ayesha",
     "age": 27, "scores": [70, 65, 72],
     "skills": ["Python", "SQL"]},
]


def validate_record(record):
    if "id" not in record:
        raise ValueError("id is required")
    if "name" not in record:
        raise ValueError("name is required")
    if record["age"] < 18 or record["age"] > 60:
        raise ValueError("age must be between 18 and 60")
    if "scores" not in record:
        raise ValueError("scores is required")
    if record["scores"] == []:
          raise ValueError("scores must not be empty")

    for score in record["scores"]:
        if score < 0 or score > 100:
            raise ValueError("scores must be between 0 and 100")

    if "skills" not in record:
        raise ValueError("skills is required")
    if record["skills"] == []:
        raise ValueError("skills must not be empty")
    return True


def process_record(record):
    total = 0
    for score in record["scores"]:
        total += score

    average = total / len(record["scores"])
    high = max(record["scores"])
    low = min(record["scores"])

    total_skills_count = 0
    for skill in record["skills"]:
        total_skills_count += 1

    is_known_python = False
    is_known_sql = False

    if "Python" in record["skills"]:
        is_known_python = True
    if "SQL" in record["skills"]:
        is_known_sql = True

    if average >= 85:
        classification = "Excellent"
    elif average >= 70:
        classification = "Good"
    elif average >= 50:
        classification = "Needs Improvement"
    else:
        classification = "Poor"

    return {
        "average": average,
        "high": high,
        "low": low,
        "skills_count": total_skills_count,
        "has_python": is_known_python,
        "has_sql": is_known_sql,
        "classification": classification
    }



averages = []
processed_records = []
highest_candidate = None
highest_average = 0
names = []

for record in records:
    try:
        validate_record(record)
        processed_record = process_record(record)
        print(processed_record)
        averages.append(processed_record["average"])
        names.append(record["name"])
        processed_records.append((record["name"], processed_record))

        if processed_record["average"] > highest_average:
            highest_average = processed_record["average"]
            highest_candidate = record["name"]

    except ValueError as e:
        print(f"Error processing record {record['id']}: {e}")

if averages:
    overall_avg = sum(averages) / len(averages)
    print(f"Overall average: {overall_avg:.2f}")
    print(f"Highest candidate: {highest_candidate} ({highest_average:.2f})")
    name_avg = dict(zip(names, averages))
    print(name_avg)

    above_avg = [name for name, avg in name_avg.items() if avg >= 80]
    above_avg=dict(zip(above_avg, averages))
    print(f"Above 80: {above_avg}")
```
