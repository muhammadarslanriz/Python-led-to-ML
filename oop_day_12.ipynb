```python
class ai_training_job:
    def __init__(self, model_name, dataset_name, samples, features, target_epochs):
        if model_name.strip() == '':
            raise ValueError("model_name cannot be empty")
        if dataset_name.strip() == '':
            raise ValueError("dataset_name cannot be empty")
        if samples <= 0:
            raise ValueError("samples must be greater than 0")
        if features <= 0:
            raise ValueError("features must be greater than 0")
        if target_epochs <= 0:
            raise ValueError("target_epochs must be greater than 0")

        self.model_name = model_name
        self.dataset_name = dataset_name
        self.samples = samples
        self.features = features
        self.target_epochs = target_epochs
        self.epochs_completed = 0
        self.accuracy = None
        self.status = "not started"

        def start(self):
            self.status="running "

        def complete_epoch(self, accuracy):
            if self.epochs_completed < 0 or self.epochs_completed < 1
                raise ValueError("epochs_completed must be greater than 0")






    jobs_data = [
    {
        "model_name": "Random Forest",
        "dataset_name": "Customer_Data",
        "samples": 12000,
        "features": 18,
        "target_epochs": 5
    },
    {
        "model_name": "Neural Network",
        "dataset_name": "Image_Data",
        "samples": 50000,
        "features": 40,
        "target_epochs": 10
    },
    {
        "model_name": "Fraud Detector",
        "dataset_name": "Fraud_Data",
        "samples": 0,
        "features": 12,
        "target_epochs": 4
    }
]

jobs = []
for data in jobs_data:
    try:
        job = ai_training_job(
            data["model_name"],
            data["dataset_name"],
            data["samples"],
            data["features"],
            data["target_epochs"]
        )
        jobs.append(job)
    except ValueError as error:
        print("Skipped", data["model_name"], "->", error)

for job in jobs:
    print(job.model_name, "|", job.dataset_name, "|", job.status)
```
