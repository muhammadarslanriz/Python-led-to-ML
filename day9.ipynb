```python
from IPython.core.interactiveshell import re
def validate_record(record):
    try:
        age = int(record["age"])
        if age < 18 or age > 60:
            raise ValueError("age must be between 18 and 60")
    except ValueError:
        print("please enter integer")
        raise

    if record["scores"] == []:
        raise ValueError("score must not be empty")

    if record["skills"] == []:
        raise ValueError("skills must not be empty")

    for s in record["scores"]:
        if s < 0 or s > 100:
            raise ValueError("score must be between 0 and 100")

    return True

def process_records(record):
    if record["scores"] == []:
        raise ValueError("score must not be empty")

    total = 0
    for i in record["scores"]:
        total += i

    avg = total / len(record["scores"])
    high = max(record["scores"])
    low = min(record["scores"]) 

    if record["skills"] == []:
        raise ValueError("skills must not be empty") 

    has = 0
    for ski in record["skills"]:
        has += 1

    knows_python = "Python" in record["skills"]

    return {
        "average": avg,
        "highest": high,
        "lowest": low,
        "num_skills": has,
        "knows_python": knows_python
    }

def serch(records, name):
    for record in records:
        if record["name"] == name:
            return process_records(record)



records = [
    {"name": "Ali",   "age": 22, "scores": [80, 85, 90],       "skills": ["Python", "SQL"]},
    {"name": "Sara",  "age": 24, "scores": [92, 88, 95],       "skills": ["Python", "ML", "Pandas"]},
    {"name": "Ahmed", "age": 21, "scores": [],                 "skills": ["Java"]},
    {"name": "Usman", "age": "twenty", "scores": [70, 75, 72], "skills": ["Python"]},
    {"name": "Hamza", "age": 23, "scores": [60, 65, 58],       "skills": []},
]

for record in records:
    try:
        validate_record(record)
        print(f"{record['name']}: valid")
    except ValueError as e:
        print(f"{record['name']}: {e}")

print()

result = serch(records, "Sara")
print(result)

# -----------------------------------------------------------


def validate_candidate(candidate):
    if not isinstance(candidate["experience"], int):
        raise ValueError("experience must be an integer")

    if candidate["experience"] < 0:
        raise ValueError("experience must not be negative")

    if candidate["skills"] == []:
        raise ValueError("skills must not be empty")

    if candidate["scores"] == []:
        raise ValueError("scores must not be empty")

    for s in candidate["scores"]:
        if s < 0 or s > 100:
            raise ValueError("score must be between 0 and 100")

    return True


def calculate_features(candidate):
    total = 0
    for s in candidate["scores"]:
        total += s

    avg = total / len(candidate["scores"])
    high = max(candidate["scores"])

    skill_count = 0
    for ski in candidate["skills"]:
        skill_count += 1

    has_python = "Python" in candidate["skills"]
    has_sql    = "SQL" in candidate["skills"]
    has_ml     = "ML" in candidate["skills"]

    candidate_score = avg
    if has_python:
        candidate_score += 5
    if has_sql:
        candidate_score += 5
    if has_ml:
        candidate_score += 10
    if candidate["experience"] >= 3:
        candidate_score += 5

    return {
        "average": avg,
        "highest": high,
        "skill_count": skill_count,
        "has_python": has_python,
        "has_sql": has_sql,
        "has_ml": has_ml,
        "candidate_score": candidate_score
    }


def clasify_candidate(candidate):
    if candidate["candidate_score"] >=100:
        return "stronge"
    elif candidate["candidate_score"] >= 80:
        return "potential"
    else:
      return "rejected"




candidates = [
    {
        "name": "Ali",
        "experience": 2,
        "skills": ["Python", "SQL", "Pandas"],
        "scores": [82, 85, 88]
    },
    {
        "name": "Sara",
        "experience": 4,
        "skills": ["Python", "ML", "PyTorch", "SQL"],
        "scores": [91, 94, 89]
    },
    {
        "name": "Bilal",
        "experience": -1,
        "skills": ["Python"],
        "scores": [70, 75]
    },
    {
        "name": "Hamza",
        "experience": 1,
        "skills": [],
        "scores": [60, 65, 62]
    }
]

strong_count = 0
potential_count = 0
rejected_count = 0
python_sql_count = 0
total_candidate_score = 0
valid_count = 0
strongest_name = None
strongest_score = 0

for c in candidates:
    try:
        validate_candidate(c)
        features = calculate_features(c)
         result = clasify_candidate(features)
        print(f"{c['name']}: {features['candidate_score']:.2f} → {result}")

        valid_count += 1
        total_candidate_score += features["candidate_score"]

        if result == "stronge":
            strong_count += 1
        elif result == "potential":
            potential_count += 1
        else:
            rejected_count += 1

        if features["has_python"] and features["has_sql"]:
            python_sql_count += 1

        if features["candidate_score"] > strongest_score:
            strongest_score = features["candidate_score"]
            strongest_name = c["name"]

    except ValueError as e:
        print(f"{c['name']}: skipped — {e}")


if valid_count > 0:
    avg_candidate_score = total_candidate_score / valid_count
else:
    avg_candidate_score = 0

print()
print("=" * 40)
print("FINAL REPORT")
print("=" * 40)
print(f"Valid candidates  : {valid_count}")
print(f"Strongest         : {strongest_name} ({strongest_score:.2f})")
print(f"Average score     : {avg_candidate_score:.2f}")
print(f"Strong            : {strong_count}")
print(f"Potential         : {potential_count}")
print(f"Rejected          : {rejected_count}")
print(f"Python + SQL      : {python_sql_count}")
print("=" * 40)
```
